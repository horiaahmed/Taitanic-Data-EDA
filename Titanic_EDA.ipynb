# Visulaizaiton

###  Survival Count (Bar Chart)
#### Create a bar chart showing how many passengers survived and how many did not.
#### ● Use the Survived column.
#### ● Add labels and a title.

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
servived=sns.countplot(x='Survived',data=dff)
plt.title("Average Fare by Passenger Class")
for container in servived.containers:
    servived.bar_label(container)

plt.show()
```

Class Distribution (Pie Chart)
Make a pie chart to show the percentage of passengers in each class (Pclass).
● Use labels 1st Class, 2nd Class, 3rd Class.
● Add percentages on the chart.


```python
counts = dff['Pclass'].value_counts()
colors = sns.color_palette("pastel")
plt.pie(
    counts,
    labels=counts.index,
    autopct='%1.1f%%',
    colors=colors
)

plt.title("Pclass Distribution")

plt.show()
```

Embarkation Port (Bar Chart)
Plot a bar chart showing the number of passengers by embarkation port (Embarked).
● X-axis: Ports (C, Q, S)
● Y-axis: Count of passengers
● Add title and labels

```python
ax = sns.countplot(x='Embarked', data=dff)
plt.title("Number of Passengers by Embarkation Port")
plt.xlabel("Embarkation Port")
plt.ylabel("Count of Passengers")

for container in ax.containers:
    ax.bar_label(container)

plt.show()
```

Fare vs Age (Scatter Plot)
Draw a scatter plot with:
● Age on the X-axis
● Fare on the Y-axis
● Color points based on Survived (different colors for survived vs not).

```python
sns.scatterplot(
    x='Age',
    y='Fare',
    hue='Survived',
    data=dff
)
plt.title("Fare vs Age by Survival")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()
```

. Survival Rate by Gender (Bar Chart)
Make a bar chart comparing the survival rate of males vs females.
● Use Sex and Survived.
● Show percentage survival for each gender

```python
ax = sns.barplot(x='Sex', y='Survived', data=dff)

plt.title("Survival Rate by Gender", fontsize=14)
plt.xlabel("Gender")
plt.ylabel("Survival Rate")

for container in ax.containers:
    ax.bar_label(container, fmt='%.2f%%')

plt.show()
```
